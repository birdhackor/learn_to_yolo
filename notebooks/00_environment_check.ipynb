# Learn to YOLO：環境檢查
這本 notebook 只列出 Python 與 PyTorch 的版本，並確認 PyTorch 能不能使用 GPU，以及有沒有 git（管理程式碼版本、也能下載程式碼的工具，各節 notebook 靠它取得教材程式）；不安裝套件，也不訓練模型。

[回到教材網站](https://birdhackor.github.io/learn_to_yolo/)

Colab 替你開來執行程式的雲端機器叫 runtime（執行環境），在繁體中文介面上顯示為「執行階段」。用 CPU 的 runtime 就能執行這本 notebook；想看能不能分到 GPU，先在選單「執行階段 → 變更執行階段類型」（英文介面是「Runtime → Change runtime type」）選一個 GPU，再執行下一格。在 Colab 執行需要登入 Google 帳號；Colab 也不保證一定分得到 GPU。

輸出裡的 `cuda_available` 表示 PyTorch 能不能透過 CUDA（NVIDIA 的 GPU 運算平台）使用 GPU，`gpu` 是那張 GPU 的型號；PyTorch 找不到可用的 GPU 時，兩者分別是 `false` 與 `null`。各節實驗都只用 CPU，不需要 GPU。

```python
import json
import platform
import shutil

report = {"python": platform.python_version(), "git_available": shutil.which("git") is not None}
try:
    import torch
except ImportError:
    report["pytorch"] = "not installed; this environment check does not install packages"
else:
    report["pytorch"] = torch.__version__
    report["cuda_available"] = torch.cuda.is_available()
    report["gpu"] = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
print(json.dumps(report, ensure_ascii=False, indent=2))
```
