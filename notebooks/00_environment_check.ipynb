# Learn to YOLO：環境檢查
這是前置準備 notebook，只確認 Python／PyTorch／GPU，不訓練模型。

[回到前置規劃網站](https://birdhackor.github.io/learn_to_yolo/)

可先用 CPU；要測 GPU，在 Colab 的 Runtime → Change runtime type 選擇。執行此頁需 Google／Colab runtime；GPU 額度不保證。

```python
import json
import platform
import shutil

report = {"python": platform.python_version(), "git_available": shutil.which("git") is not None}
try:
    import torch
except ImportError:
    report["pytorch"] = "not installed; this preparation check does not install training packages"
else:
    report["pytorch"] = torch.__version__
    report["cuda_available"] = torch.cuda.is_available()
    report["gpu"] = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
print(json.dumps(report, ensure_ascii=False, indent=2))
```
